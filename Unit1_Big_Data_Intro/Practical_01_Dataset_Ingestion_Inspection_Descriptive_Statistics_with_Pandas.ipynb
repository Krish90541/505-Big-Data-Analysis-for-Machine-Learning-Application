# Practical 01: Dataset Ingestion, Inspection & Descriptive Statistics with Pandas

**UNIT 1: Introduction to Big Data and Large Dataset Handling**

**Problem Statement:** Write a Python program using Pandas to load a structured dataset (CSV/JSON), display its first 5 rows, last 5 rows, dimensions (shape), column names, and data types with info(). Generate comprehensive summary statistics of all numeric and categorical attributes using describe(include='all').

```python
import io
import numpy as np
import pandas as pd

pd.set_option("display.max_columns", None, "display.width", 200)

rnd = np.random.default_rng(123)
n = 300
seed_df = pd.DataFrame({
    "Order ID": np.arange(1001, 1001 + n),
    "Customer Name": [f"User_{i:03d}" for i in rnd.integers(1, 100, n)],
    "City": rnd.choice(["Surat", "Mumbai", "Pune", "Nagpur", "Thane", "Rajkot"], n),
    "Product Category": rnd.choice(["Electronics", "Clothing", "Groceries", "Furniture", "Beauty"], n),
    "Quantity": rnd.integers(1, 9, n),
    "Unit Price": np.round(rnd.uniform(100, 4000, n), 2),
    "Discount %": rnd.choice([0, 5, 10, 15, 20], n),
    "Order Date": pd.date_range("2025-01-20", periods=n, freq="D").strftime("%Y-%m-%d"),
})

mem_csv = io.StringIO()
seed_df.to_csv(mem_csv, index=False)
mem_csv.seek(0)
mem_json = io.StringIO(seed_df.to_json(orient="records"))

tbl = pd.read_csv(mem_csv)
print("Dataset loaded successfully!")

print(tbl.head())

print(tbl.tail())

print("Shape (rows, columns):", tbl.shape)
print("Column names:", tbl.columns.tolist())

tbl.info()

print(tbl.describe(include='all'))

from_json = pd.read_json(mem_json)
print("JSON load – shape:", from_json.shape)
print(from_json.head())
```
