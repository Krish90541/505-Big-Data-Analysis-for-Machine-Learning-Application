# Practical 15: Pretrained Sentiment Analysis with Hugging Face Transformers

**UNIT 4: Introduction to Large Language Models and Big Data Applications**

**Problem Statement:** Write a Python program to perform Sentiment Analysis using the Hugging Face Transformers library. Initialize a pretrained sentiment-analysis pipeline (pipeline('sentiment-analysis')), classify multiple user review statements into POSITIVE or NEGATIVE sentiment labels, and display the confidence prediction scores.

```python
import pandas as pd
from transformers import pipeline

clf = pipeline("sentiment-analysis",
                      model="distilbert-base-uncased-finetuned-sst-2-english")

reviews = [
    "I absolutely love this phone, the battery life is fantastic!",
    "Great value for money and excellent customer support.",
    "The movie was amazing, I would happily watch it again.",
    "The delivery was late and the product was damaged. Very disappointed.",
    "Worst purchase ever. It stopped working after two days.",
    "The food was cold and the service was terrible.",
]

analysis = clf(reviews)

summary_df = pd.DataFrame({
    "Feedback": reviews,
    "Prediction": [r["label"] for r in analysis],
    "Confidence_Score": [round(r["score"], 4) for r in analysis],
})
pd.set_option("display.max_colwidth", 80)
print(summary_df)

pos = (summary_df["Prediction"] == "POSITIVE").sum()
print(f"{pos} POSITIVE, {len(summary_df) - pos} NEGATIVE out of {len(summary_df)} reviews")
```
