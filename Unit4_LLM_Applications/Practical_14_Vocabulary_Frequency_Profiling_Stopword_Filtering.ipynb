# Practical 14: Vocabulary Frequency Profiling & Stopword Filtering

**UNIT 4: Introduction to Large Language Models and Big Data Applications**

**Problem Statement:** Write a Python program to profile vocabulary in a text corpus: clean sentences, filter out common English stopwords (e.g., 'the', 'is', 'in', 'and'), calculate word occurrence frequencies using collections.Counter, and display the top 5 most frequent keywords.

```python
import string
from collections import Counter

corpus = [
    "Retail companies track sales data in the cloud to forecast the demand.",
    "The forecast model is trained on past data and checked on recent weeks.",
    "Chunked reading lets Pandas process files that do not fit in memory.",
    "Tokenisation splits the text into words that a model can count.",
    "Big data tools like Spark save time when the datasets are huge.",
]

NOISE_WORDS = {"the", "a", "an", "is", "are", "of", "to", "for", "on", "at", "by", "in", "and", "that", "do", "not", "fit", "can", "like", "into"}

def clean(sentence: str) -> str:
    sentence = sentence.lower()
    for p in string.punctuation:
        sentence = sentence.replace(p, "")
    return " ".join(sentence.split())

token_list = []
for s in corpus:
    token_list.extend(clean(s).split())

clean_tokens = [w for w in token_list if w not in NOISE_WORDS]

print(f"Total tokens (before filtering): {len(token_list)}")
print(f"Total tokens (after  filtering): {len(clean_tokens)}")
print(f"Vocabulary size (unique words) : {len(set(clean_tokens))}")

cnt = Counter(clean_tokens)
print(cnt)

for rank, (word, count) in enumerate(cnt.most_common(4), 1):
    print(f"{rank}. {word:<10} {count}  {'#' * count}")
```
