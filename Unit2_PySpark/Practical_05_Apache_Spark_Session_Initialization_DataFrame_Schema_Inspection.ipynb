# Practical 05: Apache Spark Session Initialization & DataFrame Schema Inspection

**UNIT 2: Big Data Processing using PySpark**

**Problem Statement:** Write a Python program to install and initialize an Apache Spark session in PySpark using SparkSession.builder. Create a structured PySpark DataFrame from Python dictionaries, display its schema tree hierarchy using printSchema(), and print its rows using .show().

```python
from pyspark.sql import SparkSession, Row

sc = (SparkSession.builder
         .appName("DS505-Pr05")
         .master("local[*]")
         .config("sc.ui.showConsoleProgress", "false")
         .getOrCreate())
sc.sparkContext.setLogLevel("ERROR")

print("Spark version:", sc.version)

data = [
    {"Name": "Aryan", "Department": "IT", "City": "Surat", "Salary": 70000, "Experience": 4},
    {"Name": "Bhumi", "Department": "HR", "City": "Pune", "Salary": 53000, "Experience": 2},
    {"Name": "Chetan", "Department": "Finance", "City": "Surat", "Salary": 64000, "Experience": 6},
    {"Name": "Devika", "Department": "IT", "City": "Mumbai", "Salary": 86000, "Experience": 8},
    {"Name": "Esha", "Department": "Sales", "City": "Surat", "Salary": 45000, "Experience": 1},
]
people = sc.createDataFrame([Row(**d) for d in data])
people.printSchema()
people.show()
sc.stop()
```
