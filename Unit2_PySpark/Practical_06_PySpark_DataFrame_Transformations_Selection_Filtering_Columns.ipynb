# Practical 06: PySpark DataFrame Transformations: Selection, Filtering & Columns

**UNIT 2: Big Data Processing using PySpark**

**Problem Statement:** Write a PySpark program to execute DataFrame transformations: select specific columns using .select(), filter rows using multi-condition boolean logic (e.g., Salary > 60000 and City == 'Surat') using .filter(), create a computed column using .withColumn(), and rename a column using .withColumnRenamed().

```python
from pyspark.sql import SparkSession, Row
from pyspark.sql.functions import col

sc = (SparkSession.builder.appName("DS505-Pr06").master("local[*]")
         .config("sc.ui.showConsoleProgress", "false").getOrCreate())
sc.sparkContext.setLogLevel("ERROR")

data = [
    {"Name": "Aryan", "Department": "IT", "City": "Surat", "Salary": 70000},
    {"Name": "Bhumi", "Department": "HR", "City": "Pune", "Salary": 53000},
    {"Name": "Chetan", "Department": "Finance", "City": "Surat", "Salary": 64000},
    {"Name": "Devika", "Department": "IT", "City": "Mumbai", "Salary": 86000},
    {"Name": "Esha", "Department": "Sales", "City": "Surat", "Salary": 45000},
    {"Name": "Farhan", "Department": "IT", "City": "Surat", "Salary": 97000},
]
people = sc.createDataFrame([Row(**d) for d in data])
people.show()

people.select("Name", "City", "Salary").show()

filtered = people.filter((col("Salary") > 65000) & (col("City") == "Surat"))
filtered.show()

with_bonus = people.withColumn("Annual_Bonus", col("Salary") * 0.10)
with_bonus.show()

renamed = with_bonus.withColumnRenamed("Name", "Employee_Name")
renamed.show()
renamed.printSchema()
sc.stop()
```
