**UNIT 3: Preparing Big Data for Machine Learning**

**Problem Statement:** Write a Python program to build and compare two classification models: DecisionTreeClassifier and KNeighborsClassifier (KNN) using Scikit-Learn. Train both models on a classification dataset, evaluate test predictions, and compare their performance using Confusion Matrix and Accuracy Score.

```python
import pandas as pd
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.metrics import confusion_matrix, accuracy_score

X, y = make_classification(n_samples=1000, n_features=10, n_informative=5, n_redundant=2,
                           n_classes=2, class_sep=1.1, flip_y=0.03, random_state=21)
print("Dataset shape:", X.shape)
print("Class counts :", dict(pd.Series(y).value_counts()))

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=21, stratify=y)
print("Train:", X_train.shape, "| Test:", X_test.shape)

algos = {
    "Decision Tree": DecisionTreeClassifier(max_depth=5, random_state=21),
    "KNN (k=5)": make_pipeline(StandardScaler(), KNeighborsClassifier(n_neighbors=5)),
}

acc_table = {}
for name, model in algos.items():
    model.fit(X_train, y_train)
    predicted = model.predict(X_test)
    acc = accuracy_score(y_test, predicted)
    acc_table[name] = acc

    print(f"===== {name} =====")
    print("Confusion matrix (rows = actual, columns = predicted):")
    print(pd.DataFrame(confusion_matrix(y_test, predicted),
                         index=["Actual 0", "Actual 1"], columns=["Pred 0", "Pred 1"]))
    print(f"Accuracy: {acc:.4f}\n")

comparison = (pd.Series(acc_table, name="Accuracy").sort_values(ascending=False).to_frame())
print(comparison)
print("Best model:", comparison.index[0])
```
